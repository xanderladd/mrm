```python
import numpy as np
import torch
import torch.nn as nn
import json
import pickle
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
from typing import Dict, List, Tuple, Optional
import pandas as pd
from tqdm import tqdm

# Configure plotting
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
plt.rcParams['figure.dpi'] = 100
plt.rcParams['savefig.dpi'] = 300
plt.rcParams['font.size'] = 10
import os
os.chdir('src')

# Import custom modules
from models.mr_gnode import MRgnODE_DynamicComm
from utils import compute_r2

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Set seeds
np.random.seed(42)
torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed(42)
```

```python
# Configuration
cache_path = Path("cache/ibl_multiregion")
config_file = "configs/ibl_multiregion.json"  # Update path as needed

# Load config
with open(config_file, 'r') as f:
    config = json.load(f)

print("Loading trained model and data...")
print(f"Session: {config['ibl_params']['eid']}")
print(f"Regions: {list(config['region_mapping'].keys())}")

# Load cached data
data_file = cache_path / 'ibl_processed_data.pkl'
with open(data_file, 'rb') as f:
    cached_data = pickle.load(f)

# Extract data splits
train_data = cached_data['processed_data']['train']
val_data = cached_data['processed_data']['val']
test_data = cached_data['processed_data']['test']

# Load model
model_params = config['model_params']
model_params['num_regions'] = len(config['region_mapping'])

model = MRgnODE_DynamicComm(**model_params).to(device)
model.load_state_dict(torch.load(cache_path / 'model.pt', map_location=device))
model.eval()

# Load metadata
with open(cache_path / 'metadata.json', 'r') as f:
    metadata = json.load(f)

print(f"\n✓ Model loaded successfully")
print(f"  Training R²: {metadata['final_r2']:.3f}")
print(f"  Validation R²: {metadata['val_r2']:.3f}")
print(f"  Total parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Display data shapes
print("Data Shapes:")
for split_name, split_data in [('Train', train_data), ('Val', val_data), ('Test', test_data)]:
    print(f"\n{split_name}:")
    for key in ['region_1', 'region_2']:
        if key in split_data:
            print(f"  {key}: {split_data[key].shape}")

# Neuron assignments
neuron_info = cached_data.get('neuron_assignments', {})
print("\nNeuron-to-Region Mapping:")
for region, info in neuron_info.items():
    if "_meta" in region: continue
    print(f"  {region}: {info['n_neurons']} neurons from {info['brain_areas']}")

# Quick validation - compute test set metrics
print("\nComputing test set metrics...")
test_regions = {k: v for k, v in test_data.items() if k.startswith('region_')}
train_regions = {k: v for k, v in train_data.items() if k.startswith('region_')}

with torch.no_grad():
    test_preds = model.predict(test_regions)
    
    if isinstance(test_preds, dict):
        test_preds_concat = np.concatenate(list(test_preds.values()), axis=-1)
        test_targets_concat = np.concatenate(list(test_regions.values()), axis=-1)
    else:
        test_preds_concat = test_preds
        test_targets_concat = np.concatenate(list(test_regions.values()), axis=-1)
    
    test_mse = np.mean((test_preds_concat - test_targets_concat) ** 2)
    test_r2 = compute_r2(test_preds_concat.flatten(), test_targets_concat.flatten())

print(f"Test MSE: {test_mse:.4f}")
print(f"Test R²: {test_r2:.3f}")

# Sample visualization
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
trial_idx = 0
time_points = np.arange(test_regions['region_1'].shape[1]) * config['ibl_params']['bin_size']

for i, region in enumerate(['region_1', 'region_2']):
    # Actual
    ax = axes[i, 0]
    im = ax.imshow(test_regions[region][trial_idx].T, aspect='auto', cmap='viridis')
    ax.set_title(f'{region} - Actual')
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Neurons')
    plt.colorbar(im, ax=ax)
    
    # Predicted
    ax = axes[i, 1]
    if isinstance(test_preds, dict):
        pred_data = test_preds[region][trial_idx].T
    else:
        start_idx = i * model_params['output_dims'][i]
        end_idx = start_idx + model_params['output_dims'][i]
        pred_data = test_preds[trial_idx, :, start_idx:end_idx].T
    
    im = ax.imshow(pred_data, aspect='auto', cmap='viridis')
    ax.set_title(f'{region} - Predicted')
    ax.set_xlabel('Time (s)')
    plt.colorbar(im, ax=ax)

plt.tight_layout()
plt.suptitle('Sample Trial Reconstruction', y=1.02)
plt.show()
```

```python
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt

# Get trial types
choice = train_data['behavior']['choice']
left_trials = np.where(choice == -1)[0]
right_trials = np.where(choice == 1)[0]

print(f"Trials: Left={len(left_trials)}, Right={len(right_trials)}")

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for i, region in enumerate(['region_1', 'region_2']):
    data = train_regions[region]  # [trials, time, neurons]
    n_trials, n_time, n_neurons = data.shape
    brain_area = config['region_mapping'][region][0]
    print(region, brain_area)
    
    # Fit PCA
    pca = PCA(n_components=min(10, n_neurons))
    data_flat = data.reshape(-1, n_neurons)
    pca.fit(data_flat)
    data_pca = pca.transform(data_flat).reshape(n_trials, n_time, -1)
    
    # Plot 1: 2D PCA trajectories
    ax = axes[i, 0]
    
    if len(left_trials) > 0:
        left_traj = data_pca[left_trials].mean(axis=0)
        ax.plot(left_traj[:, 0], left_traj[:, 1], 'b-', lw=2, label='Left')
        ax.scatter(left_traj[0, 0], left_traj[0, 1], c='blue', s=100, marker='o')
        
    if len(right_trials) > 0:
        right_traj = data_pca[right_trials].mean(axis=0)
        ax.plot(right_traj[:, 0], right_traj[:, 1], 'r-', lw=2, label='Right')
        ax.scatter(right_traj[0, 0], right_traj[0, 1], c='red', s=100, marker='o')
    
    ax.set_xlabel('PC1')
    ax.set_ylabel('PC2')
    ax.set_title(f'{brain_area} - Choice Trajectories')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # Plot 2: Euclidean distance between left/right
    ax = axes[i, 1]
    
    if len(left_trials) > 0 and len(right_trials) > 0:
        left_mean = np.mean(data_pca[left_trials], axis=0)  # [time, PCs]
        right_mean = np.mean(data_pca[right_trials], axis=0)
        
        # Calculate Euclidean distance at each time point
        distances = np.linalg.norm(left_mean - right_mean, axis=1)
        time_points = np.arange(n_time) * config['ibl_params']['bin_size'] - config['ibl_params']['pre_time']
        
        ax.plot(time_points, distances, 'k-', lw=2)
        ax.fill_between(time_points, 0, distances, alpha=0.3, color='gray')
        ax.axvline(0, color='red', linestyle=':', alpha=0.5, label='Stim On')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Distance (PC units)')
        ax.set_title(f'{brain_area} - L/R Distance')
        ax.legend()
        ax.grid(True, alpha=0.3)
        
        # Print max separation
        max_dist = distances.max()
        max_time = time_points[distances.argmax()]
        print(f"\n{brain_area}: Max L/R distance = {max_dist:.2f} at t={max_time:.2f}s")
    
    # Plot 3: Variance explained
    ax = axes[i, 2]
    var_exp = pca.explained_variance_ratio_
    cumvar = np.cumsum(var_exp)
    
    ax.bar(range(len(var_exp)), var_exp, alpha=0.6, label='Individual')
    ax.plot(range(len(var_exp)), cumvar, 'ro-', lw=2, label='Cumulative')
    ax.set_xlabel('PC')
    ax.set_ylabel('Variance Explained')
    ax.set_title(f'{brain_area} - Variance')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    print(f"  First 3 PCs: {var_exp[:3].sum():.1%} variance")
    
    # Choice selectivity
    if len(left_trials) > 0 and len(right_trials) > 0:
        left_proj = data_pca[left_trials].mean(axis=1)
        right_proj = data_pca[right_trials].mean(axis=1)
        choice_axis = right_proj.mean(axis=0) - left_proj.mean(axis=0)
        
        # d-prime
        all_proj = data_pca.mean(axis=1) @ (choice_axis / np.linalg.norm(choice_axis))
        dprime = (all_proj[right_trials].mean() - all_proj[left_trials].mean()) / \
                 np.sqrt(0.5 * (all_proj[right_trials].var() + all_proj[left_trials].var()))
        print(f"  Choice d' = {dprime:.2f}")

plt.tight_layout()
plt.suptitle('PCA Analysis: 2D Trajectories & L/R Distance', y=1.02, fontsize=14)
plt.show()
```

```python
def extract_communication_dynamics(model, data, n_samples=10):
    """Extract communication channel activities during forward pass"""
    device = next(model.parameters()).device
    dt = 0.01
    
    # Prepare data
    regions = {k: v for k, v in data.items() if k.startswith('region_')}
    region_list = [regions[f'region_{i+1}'] for i in range(len(regions))]
    
    # Sample trials
    n_trials = region_list[0].shape[0]
    sample_indices = np.random.choice(n_trials, min(n_samples, n_trials), replace=False)
    
    comm_activities = []
    region_activities = []
    
    for idx in sample_indices:
        # Get trial data
        trial_data = [torch.tensor(r[idx:idx+1], dtype=torch.float32).to(device) 
                     for r in region_list]
        
        seq_len = trial_data[0].shape[1]
        max_input_dim = max(model.input_dims)
        
        # Pad inputs
        padded_inputs = []
        for i, td in enumerate(trial_data):
            if td.shape[-1] < max_input_dim:
                padding = torch.zeros(1, seq_len, max_input_dim - td.shape[-1], device=device)
                padded = torch.cat([td, padding], dim=-1)
            else:
                padded = td[..., :max_input_dim]
            padded_inputs.append(padded)
        
        batch_input = torch.stack(padded_inputs, dim=2).squeeze(0)
        
        # Initialize state
        h = torch.zeros(1, model.N_total, device=device)
        if model.hidden_state_init == 'xavier':
            nn.init.xavier_uniform_(h)
        
        trial_comm = []
        trial_regions = []
        
        # Forward pass
        for t in range(seq_len):
            h_regions, h_comm = model.partition_state(h)
            
            # Store activities
            trial_comm.append(h_comm.detach().cpu().numpy())
            trial_regions.append([hr.detach().cpu().numpy() for hr in h_regions])
            
            # Update state
            h_dot, _ = model(h, batch_input[t:t+1], timestep=t)
            h = h + dt * h_dot
        
        comm_activities.append(np.array(trial_comm))
        region_activities.append(trial_regions)
    
    return comm_activities, region_activities, sample_indices

# Extract communication dynamics
print("Extracting communication dynamics...")
comm_acts, region_acts, sample_ids = extract_communication_dynamics(model, test_data, n_samples=20)

# Analyze communication patterns
comm_strengths = []
for trial_comm in comm_acts:
    # trial_comm shape: [time, batch=1, n_channels, comm_dim]
    trial_strength = np.linalg.norm(trial_comm.squeeze(1), axis=-1)  # [time, n_channels]
    comm_strengths.append(trial_strength)

comm_strengths = np.array(comm_strengths)  # [trials, time, n_channels]

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Average communication strength over time
ax = axes[0, 0]
mean_comm = comm_strengths.mean(axis=0)  # [time, n_channels]
time_points = np.arange(mean_comm.shape[0]) * config['ibl_params']['bin_size']

# Plot each communication channel
channel_names = []
for source in range(model.num_regions):
    for target in range(model.num_regions):
        if source != target:
            channel_names.append(f'R{source+1}→R{target+1}')

for i, name in enumerate(channel_names):
    ax.plot(time_points, mean_comm[:, i], label=name, linewidth=2)

ax.set_xlabel('Time (s)')
ax.set_ylabel('Communication Strength')
ax.set_title('Average Communication Strength Over Time')
ax.axvline(0, color='gray', linestyle=':', alpha=0.5, label='Stim On')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

# Communication strength heatmap
ax = axes[0, 1]
im = ax.imshow(mean_comm.T, aspect='auto', cmap='hot', interpolation='nearest')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Communication Channel')
ax.set_yticks(range(len(channel_names)))
ax.set_yticklabels(channel_names)
ax.set_title('Communication Strength Heatmap')
plt.colorbar(im, ax=ax, label='Strength')

# Trial-to-trial variability
ax = axes[1, 0]
std_comm = comm_strengths.std(axis=0)
for i, name in enumerate(channel_names):
    ax.fill_between(time_points, 
                    mean_comm[:, i] - std_comm[:, i],
                    mean_comm[:, i] + std_comm[:, i],
                    alpha=0.3)
    ax.plot(time_points, mean_comm[:, i], label=name, linewidth=2)

ax.set_xlabel('Time (s)')
ax.set_ylabel('Communication Strength')
ax.set_title('Communication with Trial Variability (±1 STD)')
ax.axvline(0, color='gray', linestyle=':', alpha=0.5)
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

# Communication correlation matrix
ax = axes[1, 1]
comm_flat = comm_strengths.reshape(-1, comm_strengths.shape[-1])  # [trials*time, n_channels]
corr_matrix = np.corrcoef(comm_flat.T)
im = ax.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(channel_names)))
ax.set_yticks(range(len(channel_names)))
ax.set_xticklabels(channel_names, rotation=45)
ax.set_yticklabels(channel_names)
ax.set_title('Communication Channel Correlations')
plt.colorbar(im, ax=ax, label='Correlation')

plt.tight_layout()
plt.suptitle('Inter-Region Communication Analysis', y=1.02, fontsize=12)
plt.show()

# Summary statistics
print("\nCommunication Statistics:")
for i, name in enumerate(channel_names):
    channel_mean = mean_comm[:, i].mean()
    channel_max = mean_comm[:, i].max()
    peak_time = time_points[mean_comm[:, i].argmax()]
    print(f"  {name}: Mean={channel_mean:.3f}, Max={channel_max:.3f}, Peak at {peak_time:.3f}s")

# Asymmetry analysis
if len(channel_names) == 2:  # Only for 2-region case
    asymmetry = mean_comm[:, 0] - mean_comm[:, 1]  # R1→R2 minus R2→R1
    print(f"\nCommunication Asymmetry (R1→R2 - R2→R1):")
    print(f"  Mean asymmetry: {asymmetry.mean():.3f}")
    print(f"  Max asymmetry: {asymmetry.max():.3f} at {time_points[asymmetry.argmax()]:.3f}s")
    print(f"  Min asymmetry: {asymmetry.min():.3f} at {time_points[asymmetry.argmin()]:.3f}s")
```

```python
from sklearn.decomposition import PCA

def analyze_regional_dynamics(actual_data, predicted_data, region_names, n_components=3):
    """Analyze dynamics in PCA space for each region"""
    dynamics_results = {}
    
    for i, region in enumerate(region_names):
        # Get data
        if isinstance(predicted_data, dict):
            pred = predicted_data[region]
            act = actual_data[region]
        else:
            start_idx = sum(model_params['output_dims'][:i])
            end_idx = start_idx + model_params['output_dims'][i]
            pred = predicted_data[..., start_idx:end_idx]
            act = actual_data[region]
        
        # Reshape for PCA: (trials*time, neurons)
        act_flat = act.reshape(-1, act.shape[-1])
        pred_flat = pred.reshape(-1, pred.shape[-1])
        
        # Fit PCA on actual data
        pca = PCA(n_components=min(n_components, act.shape[-1]))
        pca.fit(act_flat)
        
        # Transform both
        act_pca = pca.transform(act_flat).reshape(act.shape[0], act.shape[1], -1)
        pred_pca = pca.transform(pred_flat).reshape(pred.shape[0], pred.shape[1], -1)
        
        dynamics_results[region] = {
            'actual_pca': act_pca,
            'predicted_pca': pred_pca,
            'pca': pca,
            'variance_explained': pca.explained_variance_ratio_
        }
    
    return dynamics_results

# Analyze dynamics
test_regions = {k: v for k, v in test_data.items() if k.startswith('region_')}
with torch.no_grad():
    test_preds = model.predict(test_regions)

dynamics = analyze_regional_dynamics(test_regions, test_preds, list(test_regions.keys()))

# Visualization
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for i, region in enumerate(['region_1', 'region_2']):
    dyn = dynamics[region]
    
    # Variance explained
    ax = axes[i, 0]
    ax.bar(range(len(dyn['variance_explained'])), dyn['variance_explained'])
    ax.set_xlabel('PC')
    ax.set_ylabel('Variance Explained')
    ax.set_title(f'{region} - PCA Variance')
    ax.set_xticks(range(len(dyn['variance_explained'])))
    ax.set_xticklabels([f'PC{j+1}' for j in range(len(dyn['variance_explained']))])
    
    # PC space trajectories (first 3 trials)
    ax = axes[i, 1]
    for trial in range(min(3, dyn['actual_pca'].shape[0])):
        ax.plot(dyn['actual_pca'][trial, :, 0], dyn['actual_pca'][trial, :, 1], 
                'b-', alpha=0.5, label='Actual' if trial==0 else '')
        ax.plot(dyn['predicted_pca'][trial, :, 0], dyn['predicted_pca'][trial, :, 1], 
                'r--', alpha=0.5, label='Predicted' if trial==0 else '')
        # Mark start
        ax.scatter(dyn['actual_pca'][trial, 0, 0], dyn['actual_pca'][trial, 0, 1], 
                  c='green', s=50, marker='o', zorder=5)
    
    ax.set_xlabel('PC1')
    ax.set_ylabel('PC2')
    ax.set_title(f'{region} - Neural Trajectories')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # Speed in PC space
    ax = axes[i, 2]
    actual_speed = np.sqrt(np.sum(np.diff(dyn['actual_pca'], axis=1)**2, axis=-1))
    pred_speed = np.sqrt(np.sum(np.diff(dyn['predicted_pca'], axis=1)**2, axis=-1))
    
    time_points = np.arange(actual_speed.shape[1]) * config['ibl_params']['bin_size']
    ax.plot(time_points, actual_speed.mean(axis=0), 'b-', label='Actual', linewidth=2)
    ax.plot(time_points, pred_speed.mean(axis=0), 'r--', label='Predicted', linewidth=2)
    ax.fill_between(time_points, 
                    actual_speed.mean(axis=0) - actual_speed.std(axis=0),
                    actual_speed.mean(axis=0) + actual_speed.std(axis=0),
                    alpha=0.2, color='blue')
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Speed in PC space')
    ax.set_title(f'{region} - Trajectory Speed')
    ax.axvline(0, color='gray', linestyle=':', alpha=0.5)
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.suptitle('Regional Dynamics Analysis', y=1.02, fontsize=12)
plt.show()

# Print dynamics summary
print("Regional Dynamics Summary:")
for region in ['region_1', 'region_2']:
    dyn = dynamics[region]
    print(f"\n{region}:")
    print(f"  Variance explained by first 3 PCs: {dyn['variance_explained'][:3].sum():.3f}")
    print(f"  Effective dimensionality: {np.sum(dyn['variance_explained'] > 0.01)}")
```

```python
# Extract behavioral variables
behavior = test_data['behavior']
choice = behavior['choice']
contrast_left = behavior['stimulus_contrast_left']
contrast_right = behavior['stimulus_contrast_right']
feedback = behavior['feedback_type']

# Define trial types
left_trials = np.where(choice == -1)[0]
right_trials = np.where(choice == 1)[0]
correct_trials = np.where(feedback == 1)[0]
incorrect_trials = np.where(feedback == -1)[0]

print(f"Trial counts:")
print(f"  Left choices: {len(left_trials)}")
print(f"  Right choices: {len(right_trials)}")
print(f"  Correct: {len(correct_trials)}")
print(f"  Incorrect: {len(incorrect_trials)}")
time_points = np.arange(region_data[left_trials].shape[1])
# Analyze neural responses by trial type
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for i, region in enumerate(['region_1', 'region_2']):
    region_data = test_regions[region]
    
    # Average activity by choice
    ax = axes[i, 0]
    if len(left_trials) > 0:
        left_mean = region_data[left_trials].mean(axis=(0, 2))
        ax.plot(time_points, left_mean, 'b-', label=f'Left (n={len(left_trials)})', linewidth=2)
    if len(right_trials) > 0:
        right_mean = region_data[right_trials].mean(axis=(0, 2))
        ax.plot(time_points, right_mean, 'r-', label=f'Right (n={len(right_trials)})', linewidth=2)
    
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Mean Activity')
    ax.set_title(f'{region} - Activity by Choice')
    ax.axvline(0, color='gray', linestyle=':', alpha=0.5, label='Stim On')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # Choice selectivity
    ax = axes[i, 1]
    if len(left_trials) > 0 and len(right_trials) > 0:
        selectivity = (region_data[right_trials].mean(axis=0) - 
                      region_data[left_trials].mean(axis=0))
        im = ax.imshow(selectivity.T, aspect='auto', cmap='RdBu_r', 
                      extent=[time_points[0], time_points[-1], 0, selectivity.shape[1]])
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Neurons')
        ax.set_title(f'{region} - Choice Selectivity (R-L)')
        ax.axvline(0, color='black', linestyle=':', alpha=0.5)
        plt.colorbar(im, ax=ax)
    
    # Outcome modulation
    ax = axes[i, 2]
    if len(correct_trials) > 0:
        correct_mean = region_data[correct_trials].mean(axis=(0, 2))
        ax.plot(time_points, correct_mean, 'g-', label=f'Correct (n={len(correct_trials)})', linewidth=2)
    if len(incorrect_trials) > 0:
        incorrect_mean = region_data[incorrect_trials].mean(axis=(0, 2))
        ax.plot(time_points, incorrect_mean, 'r-', label=f'Error (n={len(incorrect_trials)})', linewidth=2)
    
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Mean Activity')
    ax.set_title(f'{region} - Activity by Outcome')
    ax.axvline(0, color='gray', linestyle=':', alpha=0.5)
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.suptitle('Trial-Type Neural Responses', y=1.02, fontsize=12)
plt.show()

# Communication differences by trial type
if len(comm_acts) > 0:
    print("\nCommunication by Trial Type:")
    
    # Map sample indices to trial types
    left_samples = [i for i, idx in enumerate(sample_ids) if idx in left_trials]
    right_samples = [i for i, idx in enumerate(sample_ids) if idx in right_trials]
    
    if left_samples and right_samples:
        left_comm = comm_strengths[left_samples].mean(axis=0)
        right_comm = comm_strengths[right_samples].mean(axis=0)
        
        for j, name in enumerate(channel_names):
            diff = right_comm[:, j].mean() - left_comm[:, j].mean()
            print(f"  {name} difference (R-L): {diff:.4f}")
```

```python
def analyze_communication_flow(model, data, n_samples=5):
    """Analyze flow fields in communication space"""
    device = next(model.parameters()).device
    
    # Sample fewer trials for detailed analysis
    regions = {k: v for k, v in data.items() if k.startswith('region_')}
    n_trials = regions['region_1'].shape[0]
    sample_indices = np.random.choice(n_trials, min(n_samples, n_trials), replace=False)
    
    flows = []
    for idx in sample_indices:
        trial_flows = []
        # Similar to extract_communication_dynamics but capture flow vectors
        # This is simplified - you'd need full flow field computation
        trial_flows.append(np.random.randn(70, model.num_comm_channels, model.comm_dim) * 0.01)
        flows.append(np.array(trial_flows))
    
    return flows

# Simplified flow visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Communication trajectory in 2D projection
ax = axes[0]
if len(comm_acts) > 0:
    # Take first trial's communication
    trial_comm = comm_acts[0].squeeze(1)  # [time, n_channels, comm_dim]
    
    # Project to 2D for visualization
    comm_flat = trial_comm.reshape(-1, trial_comm.shape[-1])
    if comm_flat.shape[0] > 2 and comm_flat.shape[1] > 1:
        pca_comm = PCA(n_components=2)
        comm_2d = pca_comm.fit_transform(comm_flat).reshape(trial_comm.shape[0], -1, 2)
        
        for ch in range(min(2, comm_2d.shape[1])):
            ax.plot(comm_2d[:, ch, 0], comm_2d[:, ch, 1], 
                   alpha=0.7, label=channel_names[ch] if ch < len(channel_names) else f'Ch{ch}')
            ax.scatter(comm_2d[0, ch, 0], comm_2d[0, ch, 1], s=50, marker='o')
            ax.scatter(comm_2d[-1, ch, 0], comm_2d[-1, ch, 1], s=50, marker='s')
    
    ax.set_xlabel('Communication PC1')
    ax.set_ylabel('Communication PC2')
    ax.set_title('Communication Trajectories (2D Projection)')
    ax.legend()
    ax.grid(True, alpha=0.3)

# Communication magnitude distribution
ax = axes[1]
if len(comm_acts) > 0:
    all_magnitudes = []
    for trial_comm in comm_acts:
        mags = np.linalg.norm(trial_comm.squeeze(1), axis=-1).flatten()
        all_magnitudes.extend(mags)
    
    ax.hist(all_magnitudes, bins=50, alpha=0.7, edgecolor='black')
    ax.axvline(np.mean(all_magnitudes), color='red', linestyle='--', 
              label=f'Mean: {np.mean(all_magnitudes):.3f}')
    ax.set_xlabel('Communication Magnitude')
    ax.set_ylabel('Count')
    ax.set_title('Distribution of Communication Strengths')
    ax.legend()

plt.tight_layout()
plt.suptitle('Communication Flow Analysis', y=1.02, fontsize=12)
plt.show()

print("Communication Flow Statistics:")
if len(comm_acts) > 0:
    print(f"  Mean magnitude: {np.mean(all_magnitudes):.4f}")
    print(f"  Std magnitude: {np.std(all_magnitudes):.4f}")
    print(f"  Max magnitude: {np.max(all_magnitudes):.4f}")
```

```python
def analyze_model_weights(model):
    """Analyze learned weights and connectivity"""
    results = {}
    
    # Communication-to-region coupling weights
    comm_to_region_weights = []
    for module in model.comm_to_region:
        weights = []
        for layer in module:
            if hasattr(layer, 'weight'):
                weights.append(layer.weight.detach().cpu().numpy())
        comm_to_region_weights.append(weights)
    
    # Gate network analysis
    gate_weights = {'regions': [], 'comm': []}
    for net in model.region_gate_nets:
        for layer in net:
            if hasattr(layer, 'weight'):
                gate_weights['regions'].append(layer.weight.detach().cpu().numpy())
                break
    
    for net in model.comm_gate_nets:
        for layer in net:
            if hasattr(layer, 'weight'):
                gate_weights['comm'].append(layer.weight.detach().cpu().numpy())
                break
    
    # Communication signs (asymmetry parameters)
    comm_signs = model.comm_signs.detach().cpu().numpy()
    
    return comm_to_region_weights, gate_weights, comm_signs

# Analyze weights
comm_weights, gate_weights, comm_signs = analyze_model_weights(model)

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Communication signs
ax = axes[0]
ax.bar(range(len(comm_signs)), comm_signs)
ax.set_xlabel('Communication Channel')
ax.set_ylabel('Sign Weight')
ax.set_title('Learned Communication Asymmetry')
ax.set_xticks(range(len(comm_signs)))
ax.set_xticklabels(channel_names)
ax.axhline(0, color='black', linestyle='-', alpha=0.3)
ax.grid(True, alpha=0.3)

# Weight magnitudes by layer type
ax = axes[1]
region_gate_mags = [np.abs(w).mean() for w in gate_weights['regions'] if len(w) > 0]
comm_gate_mags = [np.abs(w).mean() for w in gate_weights['comm'] if len(w) > 0]

x = np.arange(2)
width = 0.35
if region_gate_mags:
    ax.bar(x[0] - width/2, np.mean(region_gate_mags), width, label='Region Gates')
if comm_gate_mags:
    ax.bar(x[1] + width/2, np.mean(comm_gate_mags), width, label='Comm Gates')

ax.set_ylabel('Mean |Weight|')
ax.set_title('Gate Weight Magnitudes')
ax.set_xticks([0, 1])
ax.set_xticklabels(['Region', 'Communication'])
ax.legend()
ax.grid(True, alpha=0.3)

# Connectivity strength matrix
ax = axes[2]
connectivity = np.zeros((model.num_regions, model.num_regions))
for i, sign in enumerate(comm_signs):
    source = i // (model.num_regions - 1)
    target_offset = i % (model.num_regions - 1)
    target = target_offset if target_offset < source else target_offset + 1
    connectivity[source, target] = abs(sign)

im = ax.imshow(connectivity, cmap='hot', vmin=0)
ax.set_xlabel('Target Region')
ax.set_ylabel('Source Region')
ax.set_title('Learned Connectivity Strength')
ax.set_xticks(range(model.num_regions))
ax.set_yticks(range(model.num_regions))
ax.set_xticklabels([f'R{i+1}' for i in range(model.num_regions)])
ax.set_yticklabels([f'R{i+1}' for i in range(model.num_regions)])
plt.colorbar(im, ax=ax)

plt.tight_layout()
plt.suptitle('Model Interpretability Analysis', y=1.02, fontsize=12)
plt.show()

print("Weight Analysis Summary:")
print(f"  Communication asymmetry range: [{comm_signs.min():.3f}, {comm_signs.max():.3f}]")
print(f"  Dominant communication: {channel_names[np.argmax(np.abs(comm_signs))]}")
if region_gate_mags:
    print(f"  Mean region gate weight: {np.mean(region_gate_mags):.4f}")
if comm_gate_mags:
    print(f"  Mean comm gate weight: {np.mean(comm_gate_mags):.4f}")
```
