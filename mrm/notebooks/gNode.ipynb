```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch
import random

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

class gnODE(nn.Module):
    def __init__(self, N=2, input_dim=2, tau=0.01):
        super().__init__()
        self.N = N
        self.tau = tau
        
        # More efficient network architecture
        hidden_dim = 128  # Reduced from 316 for faster training
        
        # Flow field network F_theta
        self.flow_net = nn.Sequential(
            nn.Linear(N + input_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim), 
            nn.Tanh(),
            nn.Linear(hidden_dim, N)
        )
        
        # Gating network G_phi
        self.gate_net = nn.Sequential(
            nn.Linear(N + input_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, N),
            nn.Sigmoid()
        )
        
        # Initialize weights for better convergence
        self.apply(self._init_weights)
        
        # Move to device
        self.to(device)
    
    def _init_weights(self, module):
        """Initialize weights for faster convergence"""
        if isinstance(module, nn.Linear):
            nn.init.xavier_uniform_(module.weight, gain=0.5)
            if module.bias is not None:
                nn.init.zeros_(module.bias)
        
    def forward(self, h, x):
        # h: hidden state [batch, N]
        # x: input [batch, input_dim]
        hx = torch.cat([h, x], dim=-1)
        
        flow = self.flow_net(hx)
        gate = self.gate_net(hx)
        
        # gnODE dynamics: tau * h_dot = G(h,x) * [-h + F(h,x)]
        h_dot = gate * (-h + flow) / self.tau
        return h_dot

def euler_step(model, h, x, dt):
    """Single Euler step for ODE integration"""
    h_dot = model(h, x)
    return h + dt * h_dot

def generate_flip_flop_data(task_type, n_trials=500, trial_length=100, n_channels=2):
    """Generate data for different flip-flop tasks - optimized version"""
    data = []
    
    # Pre-allocate arrays for efficiency
    all_inputs = np.zeros((n_trials, trial_length, n_channels))
    all_targets = np.zeros((n_trials, trial_length, n_channels))
    
    for trial_idx in range(n_trials):
        # Generate pulse times (Poisson with mean 12 for 3-bit, 6 for 2-bit)
        n_pulses = np.random.poisson(6 if n_channels == 2 else 12)
        pulse_times = np.random.choice(trial_length, min(n_pulses, trial_length), replace=False)
        
        # Initialize output state
        current_output = np.zeros(n_channels)
        
        for t in range(trial_length):
            if t in pulse_times:
                if task_type == 'disk':
                    # Disk: 1 ≤ √(c1² + c2²) ≤ 2, pulses appear simultaneously
                    radius = np.random.uniform(1, 2)
                    angle = np.random.uniform(0, 2*np.pi)
                    pulse_vals = [radius * np.cos(angle), radius * np.sin(angle)]
                    all_inputs[trial_idx, t] = pulse_vals
                    current_output = pulse_vals
                else:
                    # Choose random channel for pulse
                    channel = np.random.randint(n_channels)
                    
                    if task_type == 'fixed':
                        pulse_val = np.random.choice([-1, 1])
                    elif task_type == 'square':
                        pulse_val = np.random.uniform(-1, 1)
                    elif task_type == 'rectangle':
                        if channel == 0:
                            pulse_val = np.random.uniform(-2, 2)
                        else:
                            pulse_val = np.random.uniform(-1, 1)
                    
                    all_inputs[trial_idx, t, channel] = pulse_val
                    current_output[channel] = pulse_val
            
            all_targets[trial_idx, t] = current_output.copy()
    
    # Convert to list of tuples for compatibility
    for trial_idx in range(n_trials):
        data.append((all_inputs[trial_idx], all_targets[trial_idx]))
    
    return data

def create_batched_data(train_data, batch_size=32):
    """Convert list of sequences to batched tensors"""
    inputs_list, targets_list = zip(*train_data)
    
    # Convert to tensors and batch
    inputs_batched = []
    targets_batched = []
    
    for i in range(0, len(inputs_list), batch_size):
        batch_inputs = torch.stack([torch.FloatTensor(inp) for inp in inputs_list[i:i+batch_size]]).to(device)
        batch_targets = torch.stack([torch.FloatTensor(targ) for targ in targets_list[i:i+batch_size]]).to(device)
        inputs_batched.append(batch_inputs)
        targets_batched.append(batch_targets)
    
    return inputs_batched, targets_batched

def train_gnode_efficient(model, train_data, n_epochs=600, lr=1e-3, dt=0.01, batch_size=32):
    """Efficient training with batching and optimizations"""
    optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-2)
    scheduler = optim.lr_scheduler.OneCycleLR(optimizer, max_lr=lr*10, epochs=n_epochs, 
                                            steps_per_epoch=len(train_data)//batch_size + 1)
    model.train()
    
    # Pre-batch the data
    inputs_batched, targets_batched = create_batched_data(train_data, batch_size)
    
    best_loss = float('inf')
    patience = n_epochs // 5  # Early stopping patience
    no_improve = 0
    
    for epoch in tqdm(range(n_epochs), desc="Training"):
        total_loss = 0
        
        for batch_inputs, batch_targets in zip(inputs_batched, targets_batched):
            optimizer.zero_grad()
            
            # Batch processing: [batch_size, T, input_dim]
            batch_size_actual, seq_len, input_dim = batch_inputs.shape
            
            # Initialize hidden states for all sequences in batch
            h = torch.zeros(batch_size_actual, model.N, device=device)
            outputs = []
            
            # Vectorized forward pass through time
            for t in range(seq_len):
                x_t = batch_inputs[:, t]  # [batch_size, input_dim]
                h = euler_step(model, h, x_t, dt)
                outputs.append(h)
            
            outputs = torch.stack(outputs, dim=1)  # [batch_size, T, N]
            
            # Compute loss
            loss = nn.MSELoss()(outputs, batch_targets)
            loss.backward()
            
            # Gradient clipping for stability
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            
            optimizer.step()
            scheduler.step()
            
            total_loss += loss.item()
        
        avg_loss = total_loss / len(inputs_batched)
        
        # Early stopping check
        if avg_loss < best_loss:
            best_loss = avg_loss
            no_improve = 0
        else:
            no_improve += 1
            
        if no_improve >= patience and epoch > n_epochs // 3:
            print(f"Early stopping at epoch {epoch}, best loss: {best_loss:.6f}")
            break
        
        if epoch % (n_epochs // 10) == 0:
            print(f"Epoch {epoch}, Loss: {avg_loss:.6f}, LR: {scheduler.get_last_lr()[0]:.2e}")
    
    model.eval()

# Keep the old function but make it use the new efficient one
def train_gnode(model, train_data, n_epochs=600, lr=1e-3, dt=0.01):
    """Train gnODE on flip-flop task - now uses efficient implementation"""
    return train_gnode_efficient(model, train_data, n_epochs, lr, dt, batch_size=500)

def plot_flow_field(model, task_type, xlim=(-2.5, 2.5), ylim=(-2.5, 2.5), grid_size=20):
    """Plot flow field for trained gnODE"""
    model.eval()
    
    # Create grid
    x = np.linspace(xlim[0], xlim[1], grid_size)
    y = np.linspace(ylim[0], ylim[1], grid_size)
    X, Y = np.meshgrid(x, y)
    
    # Compute flow field
    U, V = np.zeros_like(X), np.zeros_like(Y)
    
    with torch.no_grad():
        for i in range(grid_size):
            for j in range(grid_size):
                h = torch.tensor([[X[i,j], Y[i,j]]], dtype=torch.float32, device=device)
                x_input = torch.zeros(1, 2, device=device)  # No input pulse
                
                h_dot = model(h, x_input)
                U[i,j] = h_dot[0,0].cpu().item()
                V[i,j] = h_dot[0,1].cpu().item()
    
    # Plot
    plt.figure(figsize=(8, 8))
    
    # Plot flow field
    plt.quiver(X, Y, U, V, alpha=0.6, scale=50, width=0.002)
    
    # Find and plot fixed points (where flow is near zero)
    fixed_points = []
    for i in range(grid_size):
        for j in range(grid_size):
            if np.sqrt(U[i,j]**2 + V[i,j]**2) < 0.01:  # Threshold for fixed point
                fixed_points.append((X[i,j], Y[i,j]))
    
    if fixed_points:
        fixed_points = np.array(fixed_points)
        plt.scatter(fixed_points[:,0], fixed_points[:,1], c='red', s=50, zorder=5)
    
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel('Channel 1')
    plt.ylabel('Channel 2')
    plt.title(f'gnODE Flow Field - {task_type.capitalize()} Task')
    plt.grid(True, alpha=0.3)
    plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
    plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
    
    return plt.gcf()

def reproduce_figure2():
    """Reproduce Figure 2 from the paper"""
    tasks = [
        ('fixed', (-1.5, 1.5), (-1.5, 1.5)),
        ('square', (-1.5, 1.5), (-1.5, 1.5)), 
        ('rectangle', (-2.5, 2.5), (-1.5, 1.5)),
        ('disk', (-2.5, 2.5), (-2.5, 2.5))
    ]
    train_iters_lr = [[150, 0.001], [300, 0.001], [300, 0.005], [500, 0.001]]
    
    fig, axes = plt.subplots(2, 2, figsize=(12, 12))
    axes = axes.flatten()
    
    for idx, (task_type, xlim, ylim) in enumerate(tasks):
        print(f"\nTraining gnODE for {task_type} task on {device}...")
        
        # Generate data
        train_data = generate_flip_flop_data(task_type, n_trials=2000, n_channels=2)
        n_epochs, lr  = train_iters_lr[idx]
        # Create and train model
        model = gnODE(N=2, input_dim=2, tau=0.01)
        train_gnode(model, train_data, n_epochs=n_epochs, lr=lr)
        
        # Plot flow field
        plt.sca(axes[idx])
        plot_flow_field(model, task_type, xlim, ylim, grid_size=15)
        
        # Customize for each task
        if task_type == 'fixed':
            plt.title('(A) Fixed: c₁, c₂ ∈ {-1, +1}')
        elif task_type == 'square':
            plt.title('(B) Square: c₁, c₂ ∈ [-1, 1]')
        elif task_type == 'rectangle':
            plt.title('(C) Rectangle: c₁ ∈ [-2, 2], c₂ ∈ [-1, 1]')
        elif task_type == 'disk':
            plt.title('(D) Disk: 1 ≤ √(c₁² + c₂²) ≤ 2')
        
        # Clear GPU memory after each task
        torch.cuda.empty_cache() if torch.cuda.is_available() else None
    
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    # Set random seeds for reproducibility
    torch.manual_seed(42)
    np.random.seed(42)
    random.seed(42)
    
    print("Reproducing Figure 2 from gnODE paper...")
    print("This will train 4 different gnODE models and plot their flow fields.")
    reproduce_figure2()
```
